## 2. Theoretical Explanation

### 2.1 Nature of Sequential Data

Many real-world datasets consist of sequences, where each data point is
connected to the ones before it. Language, audio, weather records, and
stock prices are common examples. In these problems, understanding the
order and timing of information is essential for accurate predictions.

Traditional machine learning models treat each input independently and
therefore fail to model such temporal relationships.

---

### 2.2 Recurrent Neural Networks (RNN)

Recurrent Neural Networks are designed to process sequences by maintaining
an internal memory called the hidden state. As the model reads each element
of a sequence, it updates this hidden state using both the current input
and the previous hidden state.

This mechanism allows RNNs to retain information from earlier time steps.
However, standard RNNs struggle to learn long-term dependencies due to the
vanishing gradient problem.

---

### 2.3 Long Short-Term Memory (LSTM)

LSTM networks improve upon basic RNNs by introducing a specialized memory
cell controlled by gates. These gates regulate the flow of information and
help preserve important patterns over long sequences.

The main components are:
1. Forget Gate: removes irrelevant information.
2. Input Gate: selects new information to store.
3. Output Gate: determines what information is passed forward.

---

### 2.4 Gated Recurrent Unit (GRU)

GRU is a simplified variant of LSTM that merges some gates for efficiency.
Despite its simpler design, GRU often achieves comparable performance while
being faster to train and easier to optimize.

---

### 2.5 Convolutional Models for Sequences

Convolutional Neural Networks can also model sequential data by applying
filters along the time dimension. These filters capture local temporal
patterns, and stacking multiple layers allows the model to learn complex
long-range dependencies.

```python
from tensorflow import keras
from tensorflow.keras import layers

model = keras.Sequential([
    layers.SimpleRNN(64, return_sequences=True, input_shape=(None, 10)),
    layers.LSTM(64),
    layers.Dense(1)
])

model.summary()
```

## 3. Code Explanation

This model demonstrates how recurrent layers learn temporal relationships.
The RNN and LSTM layers process the sequence step by step, allowing the
network to capture both short-term and long-term dependencies before
producing the final output.

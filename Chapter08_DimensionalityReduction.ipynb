## 1. Chapter Summary

This chapter explains the importance of reducing the number of features in
a dataset before building machine learning models. High-dimensional data
can increase computation cost and make learning more difficult.

The chapter introduces dimensionality reduction techniques and shows how
they can improve both model performance and training efficiency.

## 2. Theoretical Explanation

### 2.1 Motivation for Dimensionality Reduction

Reducing the number of features helps simplify models, decrease training
time, and reduce the risk of overfitting. High-dimensional datasets often
contain redundant or irrelevant features that do not contribute
significantly to prediction accuracy.

---

### 2.2 Principal Component Analysis (PCA)

PCA is a popular technique that transforms the original features into a
smaller set of new features called principal components. These components
preserve as much variance as possible from the original data while
reducing dimensionality.

---

### 2.3 Other Dimensionality Reduction Methods

Other methods include Linear Discriminant Analysis (LDA), t-SNE, and
autoencoders, each of which is suitable for different types of data and
applications.

---

### 2.4 Applications of Dimensionality Reduction

Dimensionality reduction is widely used for data visualization, noise
reduction, and improving model efficiency in high-dimensional problems.

```python
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

data = load_iris()
X = data.data
y = data.target

pca = PCA(n_components=2)
X_reduced = pca.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_reduced, y, test_size=0.3, random_state=42
)

model = LogisticRegression(max_iter=200)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print("Model accuracy after PCA:", accuracy)
```

## 3. Code Explanation

This example shows how PCA can reduce the number of features while
preserving important information. After dimensionality reduction, a
classification model is trained and evaluated to demonstrate the impact on
performance.

## 4. Discussion and Conclusion

This chapter demonstrates that dimensionality reduction plays a critical
role in building efficient machine learning systems. By reducing feature
complexity, models become faster, more stable, and easier to interpret.

## 2. Theoretical Explanation

### 2.1 Basic Idea of Support Vector Machines

Support Vector Machines aim to find the best decision boundary that
separates data points of different classes. The optimal boundary is the
one that maximizes the margin between the closest data points from each
class, which are called support vectors.

---

### 2.2 Maximum Margin Principle

The maximum margin principle helps ensure that the model generalizes well
to new data. By keeping the decision boundary as far as possible from the
nearest training points, the model becomes more robust to noise and
outliers.

---

### 2.3 Kernel Trick

In many real problems, the data cannot be separated by a straight line.
The kernel trick allows SVMs to project data into higher-dimensional
spaces where a linear separation becomes possible without explicitly
computing the transformation.

---

### 2.4 Regularization and Soft Margin

Support Vector Machines allow a balance between maximizing the margin and
allowing some misclassification. This balance is controlled by a
regularization parameter that determines how much the model tolerates
classification errors.

```python
from sklearn import datasets
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

data = datasets.load_iris()
X = data.data
y = data.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = SVC(kernel='linear')
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print("Model accuracy:", accuracy)
```

## 3. Code Explanation

This example demonstrates how an SVM classifier is trained using the Iris
dataset. The model learns a decision boundary that separates the classes
and is then evaluated on unseen test data using accuracy as the
performance metric.

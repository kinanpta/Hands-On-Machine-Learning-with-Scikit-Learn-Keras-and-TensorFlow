## The Machine Learning Landscape

**Reference:**

## 1. Chapter Summary

This chapter introduces the fundamental concepts of Machine Learning and explains
what it means for a system to "learn" from data. It describes why Machine Learning
is necessary for solving complex real-world problems and highlights how ML systems
are used in modern applications such as spam filtering, recommendation systems,
speech recognition, and fraud detection.

The chapter also presents the main categories of Machine Learning systems,
including supervised learning, unsupervised learning, semi-supervised learning,
and reinforcement learning. In addition, it discusses different training strategies
such as batch learning and online learning, as well as the distinction between
instance-based learning and model-based learning.

Finally, the chapter outlines the typical workflow of a Machine Learning project
and introduces the main challenges faced in ML development, such as overfitting,
underfitting, data quality issues, and model evaluation.

## 2. Theoretical Explanation

### 2.1 What is Machine Learning?

Machine Learning is the field of study that enables computers to learn from data
without being explicitly programmed. Instead of manually writing rules, ML systems
automatically discover patterns in data and improve their performance over time.

According to the book, a program is said to learn from experience **E** with respect
to a task **T** and performance measure **P** if its performance on **T** improves
with experience **E**. This definition emphasizes that learning is measured by
performance improvement.

## 2. Theoretical Explanation

### 2.1 What is Machine Learning?

Machine Learning is a subset of artificial intelligence that enables computers to learn from data without being explicitly programmed to perform
specific tasks. A learning algorithm adapts and improves performance as it
processes more examples.

A formal definition commonly used in the field:
> A program is said to learn from experience **E** with respect to task **T** and
> performance measure **P**, if its performance at tasks in **T** improves with
> experience **E**.

This definition emphasizes three components:

1. **Task (T):** 
The system is supposed to do  
2. **Experience (E):** 
The data from which it learns  
3. **Performance (P):**
Metric to evaluate improvement

---

### 2.2 Categories of Machine Learning
Machine learning methods can be grouped based on how they learn from data, namely:
- **Supervised Learning:**
    algorithms learn from labeled data.
- **Unsupervised Learning:**
    algorithms find structure in unlabeled data.
- **Semi-Supervised Learning:**
    mix of labeled and unlabeled training data. 
- **Reinforcement Learning:**
    learner interacts with environment to maximize rewards.

---

### 2.3 Training Strategies and Concepts
Machine learning systems also differ in how they are trained, including:
- **Batch Learning:** training using the full dataset at once.  
- **Online Learning:** model updated incrementally as data arrives.
- **Instance-Based Learning:** decision based on stored examples.
- **Model-Based Learning:** generalized representation learned from data.

---

### 2.4 General Machine Learning Workflow
The general workflow in Machine Learning is as follows.
1. **Data Collection:** rather raw data.
2. **Data Preparation:** 
    clean, preprocess, transform. 
3. **Model Selection:** 
    choose algorithm class.
4. **Training:** 
    fit model to data.
5. **Evaluation:** 
    assess performance. 
6. **Tuning & Validation:**
    hyperparameter search, cross-validation.
7. **Deployment:** 
    apply model to real usage.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Create sample dataset
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([1.1, 2.0, 2.8, 3.6, 5.0])

# 2. Train linear regression model
model = LinearRegression()
model.fit(X, y)

# 3. Predictions
X_new = np.array([[6], [7]])
y_pred = model.predict(X_new)

print("Predictions:", y_pred)

# 4. Visualization
plt.scatter(X, y, color='blue', label="Training data")
plt.plot(X, model.predict(X), linestyle='--', color='red', label="Model fit")
plt.scatter(X_new, y_pred, color='green', label="Predictions")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Simple Linear Regression Example")
plt.legend()
plt.show()
```

## 3. Code Explanation

In this simple example, we demonstrate how a linear regression model can
learn from a small set of training examples. The algorithm identifies
a linear relationship between the feature (X) and target (y).

**1. Dataset**  
   A simple one-dimensional dataset with increasing values.

**2. Model Training**  
   Using `LinearRegression`, we fit a line through the training points.

**3. Prediction**  
   The model then predicts new values for unseen `X_new`.

**4. Visualization**  
   A scatter plot of training data, fitted model line, and predicted points
   help interpret model behavior visually.

## 2. Theoretical Explanation

### 2.1 Importance of Data Preprocessing

Raw data is rarely ready for direct use. Preprocessing improves model
performance by handling missing values, scaling features, and encoding
categorical variables.

---

### 2.2 Common Preprocessing Techniques

Common techniques include normalization, standardization, encoding
categorical features, and feature engineering.

---

### 2.3 Data Pipelines

Data pipelines automate preprocessing steps, ensuring consistency and
reproducibility in machine learning workflows.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

data = pd.DataFrame({
    "Age": [20, 22, 25, 30, 35],
    "Salary": [30000, 35000, 40000, 50000, 60000]
})

scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)

scaled_data
```

## 3. Code Explanation

This example demonstrates how numerical features can be scaled using
standardization, which helps many machine learning algorithms perform
better.

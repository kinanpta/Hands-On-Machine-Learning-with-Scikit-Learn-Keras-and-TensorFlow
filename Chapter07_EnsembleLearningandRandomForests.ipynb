## 1. Chapter Summary

This chapter discusses ensemble learning, a technique that combines
multiple models to achieve better performance than a single model. By
aggregating the predictions of several learners, ensemble methods reduce
errors and improve generalization.

The chapter introduces important ensemble strategies such as bagging,
boosting, and random forests, and explains why they are widely used in
practical machine learning systems.

## 2. Theoretical Explanation

### 2.1 Idea of Ensemble Learning

Ensemble learning is based on the principle that a group of diverse models
can make more accurate and reliable predictions than any individual model.
Each model contributes its own perspective, and their outputs are combined
to produce a final decision.

---

### 2.2 Bagging and Boosting

Bagging, or bootstrap aggregating, trains multiple models on different
random subsets of the training data and averages their predictions.
Boosting, on the other hand, focuses on correcting the mistakes of previous
models by giving more importance to difficult training examples.

---

### 2.3 Random Forests

Random forests are an extension of bagging that use decision trees as base
learners. Each tree is trained on a random subset of data and features,
which increases model diversity and reduces overfitting.

---

### 2.4 Benefits of Ensemble Methods

Ensemble models are often more robust, more accurate, and less sensitive to
noise compared to single models.

```python
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

data = load_iris()
X = data.data
y = data.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print("Model accuracy:", accuracy)
```

## 3. Code Explanation

This example demonstrates how a random forest classifier combines many
decision trees to improve classification performance. The model aggregates
the predictions of individual trees to produce a more accurate and stable
result.

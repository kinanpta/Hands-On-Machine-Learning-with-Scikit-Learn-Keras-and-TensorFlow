## 2. Theoretical Explanation

### 2.1 Challenges in Training Deep Networks

Deep neural networks can be difficult to train due to issues such as
vanishing and exploding gradients, slow convergence, and sensitivity to
initialization.

---

### 2.2 Weight Initialization and Activation Functions

Proper weight initialization and the choice of activation function play an
important role in stabilizing training and accelerating convergence.

---

### 2.3 Regularization Techniques

Regularization methods such as dropout, early stopping, and batch
normalization are used to reduce overfitting and improve model
generalization.

---

### 2.4 Optimization Strategies

Advanced optimization algorithms such as Adam, RMSprop, and momentum-based
gradient descent help speed up training and improve model performance.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

fashion_mnist = keras.datasets.fashion_mnist
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

X_train = X_train / 255.0
X_test = X_test / 255.0

model = keras.Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(256, activation="relu"),
    layers.Dropout(0.3),
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.fit(X_train, y_train, epochs=5, validation_split=0.1)

test_loss, test_acc = model.evaluate(X_test, y_test)
print("Test accuracy:", test_acc)
```

## 3. Code Explanation

This example demonstrates how deeper neural networks can be trained using
regularization techniques such as dropout. Validation data is used to
monitor performance and prevent overfitting during training.

## 2. Theoretical Explanation

### 2.1 Understanding Language as Data

Unlike numerical data, text is unstructured and symbolic. Before a model
can process language, the text must be converted into numbers. This process
includes several steps: tokenization, vocabulary construction, mapping
tokens to integer indices, and transforming sequences to fixed lengths
through padding or truncation.

Word embeddings are then used to represent each word as a dense numerical
vector. These embeddings capture semantic relationships between words,
allowing models to recognize similarities such as "king" and "queen" or
"good" and "excellent".

---

### 2.2 Recurrent Neural Networks for Language Modeling

Recurrent Neural Networks process text sequentially, word by word. At each
step, the model updates an internal memory called the hidden state, which
stores information from previous words. This enables the network to model
grammar, sentence structure, and contextual meaning.

However, basic RNNs struggle to preserve information over long sequences due
to the vanishing gradient problem. This makes it difficult to learn
relationships between distant words in long sentences.

---

### 2.3 Long-Term Dependencies and LSTM Networks

LSTM networks solve this limitation by introducing a memory cell controlled
by three gates: forget, input, and output. These gates regulate the flow of
information, allowing important context to be preserved while irrelevant
details are discarded.

This mechanism enables LSTMs to learn long-range dependencies in language,
such as subject–verb agreement and semantic consistency across sentences.

---

### 2.4 Attention Mechanism

The attention mechanism further improves sequence models by allowing the
network to dynamically focus on the most relevant parts of the input
sentence when generating each output. Instead of relying only on a single
compressed memory vector, the model can selectively attend to different
words in the input.

This significantly improves translation quality and contextual
understanding, especially for long and complex sentences.

---

### 2.5 Benefits of Attention in NLP

Attention improves model performance, training stability, and
interpretability. By observing attention weights, researchers can analyze
which words the model considers most important for its predictions, making
deep learning systems more transparent and easier to debug.

```python
from tensorflow import keras
from tensorflow.keras import layers

model = keras.Sequential([
    layers.Embedding(input_dim=10000, output_dim=64),
    layers.LSTM(64, return_sequences=True),
    layers.Attention(),
    layers.GlobalAveragePooling1D(),
    layers.Dense(1, activation="sigmoid")
])

model.summary()
```

## 3. Code Explanation

This model converts text into embeddings, processes the sequence using an
LSTM network, and applies an attention mechanism to highlight the most
important parts of the sentence before making a prediction. This structure
reflects modern NLP system design.

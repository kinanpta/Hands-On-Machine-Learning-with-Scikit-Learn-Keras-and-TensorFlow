## 2. Theoretical Explanation

### 2.1 Representation Learning

In traditional machine learning, feature extraction is often performed
manually. Engineers design features based on domain knowledge, which is
time-consuming and error-prone. Deep learning replaces this process with
**representation learning**, where the model automatically discovers the
most informative features from the data itself.

Lower layers of a neural network typically learn simple patterns such as
edges or basic shapes, while higher layers learn increasingly abstract
concepts such as objects, textures, or semantic meaning. This hierarchical
learning structure allows models to generalize well and perform complex
tasks with minimal manual intervention.

---

### 2.2 Autoencoders

Autoencoders are neural networks designed to learn compressed
representations of data. They consist of two parts: an **encoder**, which
maps the input into a lower-dimensional latent representation, and a
**decoder**, which reconstructs the original input from that compressed
form.

By forcing the network to reconstruct the input from limited information,
autoencoders learn essential features of the data. They are commonly used
for dimensionality reduction, noise removal, feature extraction, and data
compression.

---

### 2.3 Variational Autoencoders (VAE)

Variational Autoencoders extend basic autoencoders by learning a
probabilistic representation of the data. Instead of encoding each input as
a single point, VAEs learn a probability distribution in latent space. This
allows the model to generate new samples by sampling from that
distribution.

VAEs provide smooth and continuous latent spaces, which is particularly
useful for generative tasks and data exploration.

---

### 2.4 Generative Adversarial Networks (GAN)

GANs consist of two competing neural networks: a **generator** and a
**discriminator**. The generator attempts to produce realistic samples,
while the discriminator tries to distinguish between real data and
generated data.

Through this adversarial process, both networks improve simultaneously.
Eventually, the generator learns to create highly realistic samples. GANs
are widely used for image synthesis, super-resolution, deepfake creation,
and artistic content generation.

---

### 2.5 Applications of Generative Learning

Generative models are applied in many fields including:
- image generation and enhancement.
- data augmentation for training.
- anomaly and fraud detection.
- medical imaging analysis.
- creative content generation such as music and art.

```python
from tensorflow import keras
from tensorflow.keras import layers

encoder = keras.Sequential([
    layers.Dense(128, activation="relu"),
    layers.Dense(64, activation="relu")
])

decoder = keras.Sequential([
    layers.Dense(128, activation="relu"),
    layers.Dense(784, activation="sigmoid")
])

autoencoder = keras.Sequential([encoder, decoder])
autoencoder.summary()
```

## 3. Code Explanation

This example demonstrates a simple autoencoder architecture. The encoder
compresses the input into a lower-dimensional representation, while the
decoder reconstructs the original input. Through this process, the model
learns the most important features needed to represent the data.

## 2. Theoretical Explanation

### 2.1 Why Use Custom Models

Custom models provide greater flexibility when standard architectures are
not sufficient. They allow developers to design models tailored to
specific problems and experimental setups.

---

### 2.2 Custom Layers and Loss Functions

TensorFlow makes it possible to define custom layers and loss functions by
extending existing classes and implementing forward computation manually.

---

### 2.3 Custom Training Loops

Custom training loops provide full control over the training process. They
allow manual computation of gradients and parameter updates, enabling
advanced optimization strategies.

---

### 2.4 Advantages of Low-Level Control

Lower-level control allows fine-grained optimization, debugging, and
implementation of complex research ideas that are not supported by
standard APIs.

```python
import tensorflow as tf
from tensorflow import keras

class CustomModel(keras.Model):
    def __init__(self):
        super().__init__()
        self.dense1 = keras.layers.Dense(64, activation="relu")
        self.dense2 = keras.layers.Dense(10, activation="softmax")

    def call(self, inputs):
        x = self.dense1(inputs)
        return self.dense2(x)

model = CustomModel()

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

(X_train, y_train), (X_test, y_test) = keras.datasets.fashion_mnist.load_data()
X_train = X_train.reshape(-1, 28*28) / 255.0
X_test = X_test.reshape(-1, 28*28) / 255.0

model.fit(X_train, y_train, epochs=3)
test_loss, test_acc = model.evaluate(X_test, y_test)
print("Test accuracy:", test_acc)
```

## 3. Code Explanation

This example demonstrates how a custom neural network model is created by
subclassing the Keras Model class. This approach provides flexibility in
defining network structure and behavior.

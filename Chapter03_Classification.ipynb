## 2. Theoretical Explanation

### 2.1 What is Classification?

Classification is a type of supervised learning where the objective is to
predict the class label of an input sample. During training, the model
learns the relationship between input features and their corresponding
labels using examples that have already been labeled.

Once the model has learned these patterns, it can be used to predict the
class of new and unseen data.

---

### 2.2 Types of Classification

Classification problems can be divided into several categories.

1. **Binary Classification:**  
   the model predicts between two possible classes, such as true or false.

2. **Multiclass Classification:**  
   the model chooses from more than two classes, such as recognizing
   handwritten digits.

3. **Multilabel Classification:**  
   a single input can belong to more than one class at the same time.

---

### 2.3 Evaluation Metrics for Classification

Evaluating a classification model is essential to understand how well it
performs on unseen data.

1. **Accuracy:**  
   measures how many predictions are correct.

2. **Precision:**  
   shows how reliable the positive predictions are.

3. **Recall:**  
   indicates how many actual positive samples are detected.

4. **Confusion Matrix:**  
   provides a detailed summary of correct and incorrect predictions.

---

### 2.4 Improving Classification Models

Classification models can be improved through several techniques.

1. **Feature Engineering:**  
   improving the representation of input data.

2. **Algorithm Selection:**  
   choosing models that are suitable for the problem.

3. **Hyperparameter Tuning:**  
   adjusting model parameters to achieve better performance.

4. **Cross-Validation:**  
   ensuring stable and reliable evaluation results.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

data = load_iris()
X = data.data
y = data.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = LogisticRegression(max_iter=200)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print("Model accuracy:", accuracy)
```

## 3. Code Explanation

This example demonstrates a simple classification workflow using the Iris
dataset. The dataset is first divided into training and testing sets. A
logistic regression model is then trained on the training data and used to
predict the class labels of unseen samples. The accuracy score measures
how well the model performs.

## 2. Theoretical Explanation

### 2.1 Structure of Neural Networks

Artificial neural networks are inspired by the structure of the human
brain. They consist of layers of interconnected nodes called neurons. Each
connection has a weight that is adjusted during training.

---

### 2.2 Activation Functions

Activation functions introduce non-linearity into the network. Common
activation functions include ReLU, sigmoid, and softmax, which allow the
network to learn complex patterns in data.

---

### 2.3 Training Neural Networks

Training a neural network involves adjusting its weights using
optimization algorithms such as gradient descent. The goal is to minimize
a loss function that measures the difference between predictions and
actual values.

---

### 2.4 Role of Keras

Keras provides a simple and efficient interface for building, training,
and evaluating neural networks on top of TensorFlow.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Load dataset
fashion_mnist = keras.datasets.fashion_mnist
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

# Normalize data
X_train = X_train / 255.0
X_test = X_test / 255.0

# Build model
model = keras.Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Train model
model.fit(X_train, y_train, epochs=5)

# Evaluate model
test_loss, test_acc = model.evaluate(X_test, y_test)
print("Test accuracy:", test_acc)
```

## 3. Code Explanation

This example demonstrates how a basic neural network can be built and
trained using Keras. The Fashion MNIST dataset is used to classify clothing
images. The network learns features from pixel values and improves its
accuracy through training.

## 1. Chapter Summary

This chapter introduces Convolutional Neural Networks (CNNs), which are the
core technology behind modern computer vision systems. CNNs are designed
to automatically extract meaningful features from images, making them
highly effective for tasks such as image classification, object detection,
and facial recognition.

The chapter explains the structure of CNNs, including convolutional layers,
pooling layers, and fully connected layers, and discusses why CNNs perform
significantly better than traditional machine learning models on image
data.

## 2. Theoretical Explanation

### 2.1 Motivation for Convolutional Neural Networks

Images contain strong spatial relationships. Nearby pixels are often more
related than distant ones. Traditional neural networks ignore this
structure, while CNNs are specifically designed to exploit it using local
connections and shared weights.

---

### 2.2 Convolutional Layers

Convolutional layers apply small filters that slide across the input image.
Each filter detects specific patterns such as edges, textures, or shapes.
These layers allow the network to gradually build complex representations
from simple visual features.

---

### 2.3 Pooling Layers

Pooling layers reduce the spatial size of feature maps, making the network
more computationally efficient and helping it become more robust to small
image shifts and distortions.

---

### 2.4 Fully Connected Layers

After feature extraction, fully connected layers combine the learned
features to perform final classification or prediction.

```python
from tensorflow import keras
from tensorflow.keras import layers

model = keras.Sequential([
    layers.Conv2D(32, (3, 3), activation="relu", input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")
])

model.summary()
```

## 3. Code Explanation

This example shows the basic structure of a convolutional neural network.
The convolutional layers extract visual features, pooling layers reduce
dimensionality, and fully connected layers perform classification based on
the extracted features.
